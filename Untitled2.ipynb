```python

import numpy as np
import pandas as pd
import seaborn as sns

# SETUP: Creating dummy data for demonstration

np.random.seed(42)

data1 = pd.DataFrame(
    {
        "ID": range(1, 101),
        "Age": np.random.choice(
            [25, 30, 35, 40, np.nan, 50], size=100
        ),
        "Salary": np.random.normal(50000, 15000, 100),
        "Purchased": np.random.choice(["Yes", "No"], size=100),
    }
)

# Duplicate rows to clean later
data1 = pd.concat(
    [data1, data1.iloc[[0, 1]]], ignore_index=True
)

data2 = pd.DataFrame(
    {
        "ID": range(1, 101),
        "Score": np.random.randint(50, 100, 100),
        "Redundant_Col": np.random.rand(100),
    }
)

# 1. EXPLORATORY DATA ANALYSIS (EDA)

print("--- 1. EXPLORATORY DATA ANALYSIS (EDA) ---")

print(data1.info())

print("\nSummary statistics:")
print(data1.describe())

print("\nValue counts:")
print(data1["Purchased"].value_counts())



```

```python



# 2. DATA CLEANING & MISSING VALUE TREATMENT

print("\n--- 2. DATA CLEANING ---")

print("Missing values before treatment:")
print(data1.isnull().sum())

# Remove duplicate rows
data1.drop_duplicates(inplace=True)

# Fill missing Age values with median
age_median = data1["Age"].median()

data1["Age"] = data1["Age"].fillna(age_median)

print("Missing values after treatment:")
print(data1["Age"].isnull().sum())


```

```python
# 4. DATA INTEGRATION

print("\n--- 4. DATA INTEGRATION ---")

df_integrated = pd.merge(
    data1, data2, on="ID", how="inner"
)

print("Integrated dataset shape:", df_integrated.shape)
```

```python
# 5. DATA REDUCTION

print("\n--- 5. DATA REDUCTION ---")

df_reduced = df_integrated.drop(
    columns=["Redundant_Col"]
)

print("Reduced dataset shape:", df_reduced.shape)
```

```python

# 6. CORRELATION ANALYSIS

print("\n--- 6. CORRELATION ANALYSIS ---")

numeric_cols = df_reduced.select_dtypes(
    include=[np.number]
).drop(columns=["ID"])

correlation_matrix = numeric_cols.corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

```python

import numpy as np
import pandas as pd
import seaborn as sns

# Creating dummy data
np.random.seed(42)

data1 = pd.DataFrame({
    "ID": range(1, 101),
    "Age": np.random.choice(
        [25, 30, 35, 40, np.nan, 50], size=100
    ),
    "Salary": np.random.normal(50000, 15000, 100),
    "Purchased": np.random.choice(["Yes", "No"], size=100)
})

# Duplicate rows
data1 = pd.concat(
    [data1, data1.iloc[[0, 1]]],
    ignore_index=True
)

data2 = pd.DataFrame({
    "ID": range(1, 101),
    "Score": np.random.randint(50, 100, 100),
    "Redundant_Col": np.random.rand(100)
})

# 1. EDA AND SUMMARY STATISTICS
print("--- 1. EDA & SUMMARY STATISTICS ---")

print(data1.info())

print("\nSummary Statistics:")
print(data1.describe())

print("\nValue Counts:")
print(data1["Purchased"].value_counts())


# 2. DATA CLEANING
print("\n--- 2. DATA CLEANING ---")

print("Missing Values Before Treatment:")
print(data1.isnull().sum())

# Remove duplicates
data1.drop_duplicates(inplace=True)

# Fill missing values with median
age_median = data1["Age"].median()

data1["Age"].fillna(age_median, inplace=True)

print("Missing Values After Treatment:")
print(data1["Age"].isnull().sum())


# 3. DATA TRANSFORMATION
print("\n--- 3. DATA TRANSFORMATION ---")

# Min-Max Scaling
data1["Salary_Normalized"] = (
    (data1["Salary"] - data1["Salary"].min()) /
    (data1["Salary"].max() - data1["Salary"].min())
)

# One-Hot Encoding
data1 = pd.get_dummies(
    data1,
    columns=["Purchased"],
    drop_first=True
)

print(data1[
    ["Age", "Salary_Normalized", "Purchased_Yes"]
].head())


# 4. DATA INTEGRATION
print("\n--- 4. DATA INTEGRATION ---")

df_integrated = pd.merge(
    data1,
    data2,
    on="ID",
    how="inner"
)

print("Integrated Dataset Shape:")
print(df_integrated.shape)


# 5. DATA REDUCTION
print("\n--- 5. DATA REDUCTION ---")

df_reduced = df_integrated.drop(
    columns=["Redundant_Col"]
)

print("Reduced Dataset Shape:")
print(df_reduced.shape)


# 6. CORRELATION ANALYSIS
print("\n--- 6. CORRELATION ANALYSIS ---")

numeric_cols = df_reduced.select_dtypes(
    include=[np.number]
).drop(columns=["ID"])

correlation_matrix = numeric_cols.corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

```python

import numpy as np
import pandas as pd

# 1. DATA INTEGRATION

# Creating two datasets

df_demographics = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "age": [25, np.nan, 47, 31],
    "city": ["NY", "LA", "SF", "NY"]
})

df_financials = pd.DataFrame({
    "id": [1, 2, 3, 5],
    "income": [70000, 85000, np.nan, 95000]
})

# Merge datasets
df = pd.merge(
    df_demographics,
    df_financials,
    on="id",
    how="outer"
)

print("--- After Data Integration ---")
print(df)


# 2. DATA CLEANING AND MISSING VALUE TREATMENT

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Fill missing values
df["age"] = df["age"].fillna(df["age"].median())

df["income"] = df["income"].fillna(
    df["income"].median()
)

df["city"] = df["city"].fillna("Unknown")

print("\n--- After Missing Value Treatment ---")
print(df)


# 3. DATA TRANSFORMATION

# Min-Max Scaling
df["income_scaled"] = (
    (df["income"] - df["income"].min()) /
    (df["income"].max() - df["income"].min())
)

# One-Hot Encoding
df = pd.get_dummies(
    df,
    columns=["city"],
    dtype=int
)

print("\n--- After Data Transformation ---")
print(df)


# 4. DATA REDUCTION

df_reduced = df.drop(columns=["income"])

print("\n--- After Data Reduction ---")
print(df_reduced)


# 5. EDA AND SUMMARY STATISTICS

print("\n--- Summary Statistics ---")
print(df_reduced.describe())

print("\n--- Data Structure & Information ---")
print(df_reduced.info())


# 6. CORRELATION ANALYSIS

print("\n--- Correlation Matrix ---")

correlation_matrix = df_reduced.corr()

print(correlation_matrix)
```
