```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Creating dummy data
np.random.seed(42)

data1 = pd.DataFrame({
    "ID": range(1, 101),
    "Age": np.random.choice([25, 30, 35, 40, np.nan, 50], 100),
    "Salary": np.random.normal(50000, 15000, 100),
    "Purchased": np.random.choice(["Yes", "No"], 100)
})

# Adding duplicate rows
data1 = pd.concat(
    [data1, data1.iloc[[0, 1]]],
    ignore_index=True
)

data2 = pd.DataFrame({
    "ID": range(1, 101),
    "Score": np.random.randint(50, 100, 100),
    "Redundant_Col": np.random.rand(100)
})

# 2. EDA
print("--- EDA ---")
print(data1.info())
print(data1.describe())
print(data1["Purchased"].value_counts())

purchase_counts = data1["Purchased"].value_counts()

# 3. Data Cleaning
print("\n--- Data Cleaning ---")
print(data1.isnull().sum())

data1 = data1.drop_duplicates().copy()

age_median = data1["Age"].median()
data1["Age"] = data1["Age"].fillna(age_median)

print(data1.isnull().sum())

# 4. Data Transformation
print("\n--- Data Transformation ---")

data1["Salary_Normalized"] = (
    (data1["Salary"] - data1["Salary"].min()) /
    (data1["Salary"].max() - data1["Salary"].min())
)

data1 = pd.get_dummies(
    data1,
    columns=["Purchased"],
    drop_first=True,
    dtype=int
)

print(data1.head())

# 5. Data Integration
print("\n--- Data Integration ---")

df_integrated = pd.merge(
    data1, data2, on="ID", how="inner"
)

print(df_integrated.shape)

# 6. Data Reduction
print("\n--- Data Reduction ---")

df_reduced = df_integrated.drop(
    columns=["Redundant_Col"]
)

print(df_reduced.shape)

# 7. Correlation Analysis
print("\n--- Correlation Analysis ---")

numeric_cols = df_reduced.select_dtypes(
    include=[np.number]
).drop(columns=["ID"])

correlation_matrix = numeric_cols.corr()

print(correlation_matrix)

# 8. Data Visualization

sns.set_theme(style="whitegrid")

# Plot 1: Age Histogram
plt.figure(figsize=(7, 4))
plt.hist(data1["Age"], bins=8, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Plot 2: Salary Histogram
plt.figure(figsize=(7, 4))
plt.hist(data1["Salary"], bins=12, edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()

# Plot 3: Pie Chart
plt.figure(figsize=(6, 6))
plt.pie(
    purchase_counts.values,
    labels=purchase_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Purchase Distribution")
plt.show()

# Plot 4: Bar Chart
average_salary = [
    data1.loc[data1["Purchased_Yes"] == 0, "Salary"].mean(),
    data1.loc[data1["Purchased_Yes"] == 1, "Salary"].mean()
]

plt.figure(figsize=(7, 4))
plt.bar(["No", "Yes"], average_salary)
plt.title("Average Salary by Purchase Category")
plt.xlabel("Purchased")
plt.ylabel("Average Salary")
plt.show()

# Plot 5: Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# SETUP: Creating dummy data
np.random.seed(42)

data1 = pd.DataFrame({
    "ID": range(1, 101),
    "Age": np.random.choice(
        [25, 30, 35, 40, np.nan, 50], size=100
    ),
    "Salary": np.random.normal(50000, 15000, 100),
    "Purchased": np.random.choice(["Yes", "No"], size=100)
})

# Duplicate rows
data1 = pd.concat(
    [data1, data1.iloc[[0, 1]]],
    ignore_index=True
)

data2 = pd.DataFrame({
    "ID": range(1, 101),
    "Score": np.random.randint(50, 100, 100),
    "Redundant_Col": np.random.rand(100)
})

# 1. EDA
print("--- 1. EDA & SUMMARY STATISTICS ---")
print(data1.info())
print(data1.describe())
print(data1["Purchased"].value_counts())

# 2. DATA CLEANING
print("\n--- 2. DATA CLEANING ---")
print(data1.isnull().sum())

data1.drop_duplicates(inplace=True)

age_median = data1["Age"].median()
data1["Age"] = data1["Age"].fillna(age_median)

print("Missing Age after treatment:",
      data1["Age"].isnull().sum())

# 3. DATA TRANSFORMATION
print("\n--- 3. DATA TRANSFORMATION ---")

data1["Salary_Normalized"] = (
    (data1["Salary"] - data1["Salary"].min()) /
    (data1["Salary"].max() - data1["Salary"].min())
)

data1 = pd.get_dummies(
    data1,
    columns=["Purchased"],
    drop_first=True,
    dtype=int
)

print(data1[
    ["Age", "Salary_Normalized", "Purchased_Yes"]
].head())

# 4. DATA INTEGRATION
print("\n--- 4. DATA INTEGRATION ---")

df_integrated = pd.merge(
    data1, data2, on="ID", how="inner"
)

print("Integrated dataset shape:",
      df_integrated.shape)

# 5. DATA REDUCTION
print("\n--- 5. DATA REDUCTION ---")

df_reduced = df_integrated.drop(
    columns=["Redundant_Col"]
)

print("Reduced dataset shape:",
      df_reduced.shape)

# 6. CORRELATION ANALYSIS
print("\n--- 6. CORRELATION ANALYSIS ---")

numeric_cols = df_reduced.select_dtypes(
    include=[np.number]
).drop(columns=["ID"])

correlation_matrix = numeric_cols.corr()

print("Correlation Matrix:")
print(correlation_matrix)

# 7. GRAPHICAL VISUALIZATION
sns.set_theme(style="whitegrid")

# Plot 1: Age Histogram
plt.figure(figsize=(7, 4))
plt.hist(data1["Age"], bins=8, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Plot 2: Salary Histogram
plt.figure(figsize=(7, 4))
plt.hist(data1["Salary"], bins=12, edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()

# Plot 3: Purchase Pie Chart
purchase_counts = data1["Purchased_Yes"].map(
    {0: "No", 1: "Yes"}
).value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    purchase_counts.values,
    labels=purchase_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Purchase Distribution")
plt.show()

# Plot 4: Average Salary Bar Chart
average_salary = data1.groupby(
    "Purchased_Yes"
)["Salary"].mean()

plt.figure(figsize=(7, 4))
plt.bar(["No", "Yes"], average_salary.values)
plt.title("Average Salary by Purchase Category")
plt.xlabel("Purchased")
plt.ylabel("Average Salary")
plt.show()

# Plot 5: Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```
